```python
import shutil
import os

folder = "/content"
for filename in os.listdir(folder):
    file_path = os.path.join(folder, filename)
    try:
        if os.path.isfile(file_path) or os.path.islink(file_path):
            os.unlink(file_path)  # 파일 또는 심볼릭 링크 삭제
        elif os.path.isdir(file_path):
            shutil.rmtree(file_path)  # 폴더(디렉토리) 삭제
    except Exception as e:
        print(f'Failed to delete {file_path}. Reason: {e}')

#위 코드는 아래 코드들에서 오류났을 때, 새로운 파일이 생기니까 그 파일을 다 삭제해주는 코드! (리셋느낌)
```

```python
model_path = "yolo11x-seg.pt"  # XL segmentation 모델 파일명
# 직접 다운로드 받아 구글 드라이브나 Colab 환경에 올려두셔야 합니다.
```

```python
import cv2
from ultralytics import solutions

video_path = filename

cap = cv2.VideoCapture(video_path) #OpenCV로 영상 파일을 읽는 객체 생성
w, h, fps = (int(cap.get(x)) for x in (cv2.CAP_PROP_FRAME_WIDTH, cv2.CAP_PROP_FRAME_HEIGHT, cv2.CAP_PROP_FPS))
#영상의 너비, 높이, 프레임 수 정보 추출, 변수 w, h, fps에 영상 크기와 초당 프레임 수 저장 (int형 변환)
video_writer = cv2.VideoWriter("output.avi", cv2.VideoWriter_fourcc(*"mp4v"), fps, (w, h))
#새 영상 저장용 VideoWriter 객체 생성

isegment = solutions.InstanceSegmentation(
    show=True,
    model=model_path
) #Ultralytics의 인스턴스 세그멘테이션 모델 불러오기

# 영상 프레임별 반복 처리 루프
while cap.isOpened():
    success, im0 = cap.read()
    if not success:
        break
    results = isegment(im0)
    video_writer.write(results.plot_im)
cap.release()
video_writer.release()
cv2.destroyAllWindows()
```
